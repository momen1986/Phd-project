```python

import numpy as np
import pandas as pd
from scipy.interpolate import interp1d

# ==========================================
# ORIGINAL SALAMN DATA
# ==========================================

pressure = np.array([
     47.86, 95.72, 145.04, 194.35, 242.21,
     339.39, 436.46, 630.91, 825.76,
     1166.10, 1841.98, 2865.95, 3933.42,
     5392.50, 6897.99
])

experimental = np.array([
    273.22, 276.88, 279.23, 281.16, 283.17,
    285.24, 287.32, 289.18, 291.32,
    293.32, 295.24, 297.24, 299.22,
    301.28, 303.19
])

hysys = np.array([
    273.91, 277.81, 281.11, 283.46, 285.26,
    287.91, 289.78, 292.28, 293.85,
    295.44, 296.82, 298.30, 299.70,
    301.40, 303.47
])

# ==========================================
# INTERPOLATION FUNCTIONS
# ==========================================

f_exp = interp1d(
    pressure,
    experimental,
    kind="cubic"
)

f_hysys = interp1d(
    pressure,
    hysys,
    kind="cubic"
)

# ==========================================
# NEW PRESSURES
# ==========================================

new_pressure = np.linspace(
    700,
    3000,
    13
)

# ==========================================
# TEMPERATURES
# ==========================================

new_exp = f_exp(new_pressure)

new_hysys = f_hysys(new_pressure)

# ==========================================
# RESULTS
# ==========================================

results = pd.DataFrame({

    "Pressure (psia)": np.round(new_pressure, 2),

    "Experimental (K)": np.round(new_exp, 2),

    "HYSYS (K)": np.round(new_hysys, 2)

})

print(results)

```

```python
import numpy as np
import pandas as pd
from scipy.interpolate import interp1d

# ==========================================
# SAMPLE 11 DATA
# ==========================================

pressure = np.array([
    150,
    250,
    350,
    500,
    700,
    900,
    1200,
    1600
])

experimental = np.array([
    274.8,
    277.6,
    280.4,
    283.2,
    286.0,
    288.7,
    291.5,
    293.2
])

# ==========================================
# INTERPOLATION
# ==========================================

f_exp = interp1d(
    pressure,
    experimental,
    kind="cubic",
    fill_value="extrapolate"
)

# ==========================================
# NEW PRESSURES
# ==========================================

new_pressure = np.linspace(
    590,
    1600,
    10
)

# ==========================================
# PREDICTED TEMPERATURES
# ==========================================

new_exp = f_exp(new_pressure)

# ==========================================
# RESULTS
# ==========================================

results = pd.DataFrame({

    "Pressure (psia)": np.round(new_pressure, 2),

    "Experimental (K)": np.round(new_exp, 2)

})

print(results)
```

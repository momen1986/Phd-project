## 2- DATA DESCRIPTION 

## 2-1 COMPOSITION DATASET

```python
import pandas as pd
import matplotlib.pyplot as plt

# ==========================================
# LOAD DATA
# ==========================================

df = pd.read_excel("composition_dataset_clean.xlsx")

# ==========================================
# FEATURES
# ==========================================

features = [
    'Water',
    'Methane',
    'Ethane',
    'Propane',
    'Isobutane',
    'Butane',
    'Isopentane',
    'Pentane',
    'Hexane',
    'Heptane',
    'Carbon Dioxide',
    'SPECIFIC GRAVITY',
    'Pressure (psia)',
    'PR',
    'Hydrate Temperature (K)'
]

# ==========================================
# SUMMARY STATISTICS TABLE
# ==========================================

summary = pd.DataFrame(index=features)

summary["Mean"] = df[features].mean()
summary["Std Dev"] = df[features].std()
summary["Min"] = df[features].min()
summary["Q1"] = df[features].quantile(0.25)
summary["Median"] = df[features].median()
summary["Q3"] = df[features].quantile(0.75)
summary["Max"] = df[features].max()

summary = summary.round(4)

print("\nSUMMARY STATISTICS")
print(summary)

summary.to_excel(
    "Composition_Summary_Statistics.xlsx"
)

# ==========================================
# HISTOGRAMS
# ==========================================

n_features = len(features)

ncols = 4
nrows = (n_features + ncols - 1) // ncols

fig, axes = plt.subplots(
    nrows,
    ncols,
    figsize=(18, 12)
)

axes = axes.flatten()

for i, col in enumerate(features):

    axes[i].hist(
        df[col],
        bins=30
    )

    axes[i].set_title(col, fontsize=9)
    axes[i].set_ylabel("Frequency")

# remove empty subplots
for j in range(i + 1, len(axes)):
    fig.delaxes(axes[j])

plt.suptitle(
    f"Composition Dataset (n = {len(df):,})",
    fontsize=16,
    fontweight='bold'
)

plt.tight_layout()

plt.show()
```

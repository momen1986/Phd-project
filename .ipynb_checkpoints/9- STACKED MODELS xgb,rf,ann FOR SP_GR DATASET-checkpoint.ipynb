```python
# ==========================================================
# CELL 1: IMPORT REQUIRED LIBRARIES
# ==========================================================
# SP_GR HYBRID STACKING
# RF + XGB + ANN
# ==========================================================

import numpy as np
import pandas as pd
import random
import joblib
import tensorflow as tf

from sklearn.model_selection import GroupKFold

from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

from xgboost import XGBRegressor

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import Dense

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import EarlyStopping

# ==========================================================
# RANDOM SEED
# ==========================================================

SEED = 42

np.random.seed(SEED)

random.seed(SEED)

tf.random.set_seed(SEED)

print("Libraries Loaded Successfully")

# ==========================================================
# CELL 2: LOAD SP_GR DATASET
# ==========================================================

df = pd.read_csv(
    "sp_gr_dataset_dispersion.csv"
)

TARGET = "Hydrate Temperature (K)"

# ==========================================================
# FEATURES
# ==========================================================

X = df[
    [
        "Specific Gravity",
        "Pressure (psia)"
    ]
]

y = df[TARGET]

# ==========================================================
# GROUPING
# SAME GAS AT DIFFERENT PRESSURES
# ==========================================================

df["group_id"] = (
    df["Specific Gravity"]
    .round(4)
    .astype(str)
)

groups = df["group_id"]

# ==========================================================
# CELL 3: TRAIN TEST SPLIT
# ==========================================================

from sklearn.model_selection import GroupShuffleSplit

gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(
        X,
        y,
        groups
    )
)

X_train = X.iloc[train_idx]
X_test  = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test  = y.iloc[test_idx]

groups_train = groups.iloc[train_idx]

print("Train Samples :", len(X_train))
print("Test Samples  :", len(X_test))

print(
    "Train Groups :",
    groups_train.nunique()
)

print(
    "Test Groups :",
    groups.iloc[test_idx].nunique()
)
```

```python
# ==========================================================
# CELL 4: DEFINE BASE LEARNERS
# ==========================================================

from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1
)

xgb = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=3,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

print("RF and XGB Initialized")
```

```python
# ==========================================================
# CELL 5: ANN MODEL
# ==========================================================

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam

def build_ann():

    model = Sequential()

    model.add(
        Dense(
            256,
            activation="relu",
            input_shape=(X_train.shape[1],)
        )
    )

    model.add(
        Dense(
            128,
            activation="relu"
        )
    )

    model.add(
        Dense(
            1,
            activation="linear"
        )
    )

    model.compile(
        optimizer=Adam(
            learning_rate=0.0005
        ),
        loss="mse"
    )

    return model
```

```python
# ==========================================================
# CELL 6: INITIALIZE GROUPKFOLD AND OOF ARRAYS
# ==========================================================
# Out-of-Fold predictions will be generated using
# GroupKFold to avoid data leakage.
# ==========================================================

from sklearn.model_selection import GroupKFold

gkf = GroupKFold(
    n_splits=5
)

# ==========================================================
# OOF ARRAYS
# ==========================================================

oof_rf = np.zeros(
    len(X_train)
)

oof_xgb = np.zeros(
    len(X_train)
)

oof_ann = np.zeros(
    len(X_train)
)

print(
    "Number of Training Samples:",
    len(X_train)
)

print(
    "Number of Groups:",
    groups_train.nunique()
)

print(
    "Number of Folds:",
    gkf.get_n_splits()
)

print(
    "OOF Arrays Created Successfully"
)
```

```python
# ==========================================================
# CELL 7: GENERATE OOF PREDICTIONS
# ==========================================================

from tensorflow.keras.callbacks import EarlyStopping

for fold, (train_idx, valid_idx) in enumerate(

    gkf.split(
        X_train,
        y_train,
        groups_train
    )

):

    print("\n" + "="*60)
    print(f"FOLD {fold+1}")
    print("="*60)

    # ======================================================
    # SPLIT
    # ======================================================

    X_tr = X_train.iloc[train_idx]
    X_va = X_train.iloc[valid_idx]

    y_tr = y_train.iloc[train_idx]
    y_va = y_train.iloc[valid_idx]

    # ======================================================
    # RF
    # ======================================================

    rf.fit(
        X_tr,
        y_tr
    )

    rf_pred = rf.predict(
        X_va
    )

    oof_rf[valid_idx] = rf_pred

    # ======================================================
    # XGB
    # ======================================================

    xgb.fit(
        X_tr,
        y_tr
    )

    xgb_pred = xgb.predict(
        X_va
    )

    oof_xgb[valid_idx] = xgb_pred

    # ======================================================
    # ANN SCALING
    # ======================================================

    X_scaler = StandardScaler()

    y_scaler = StandardScaler()

    X_tr_ann = X_scaler.fit_transform(
        X_tr
    )

    X_va_ann = X_scaler.transform(
        X_va
    )

    y_tr_ann = y_scaler.fit_transform(
        y_tr.values.reshape(-1,1)
    )

    # ======================================================
    # ANN
    # ======================================================

    ann = build_ann()

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=50,
        restore_best_weights=True,
        verbose=0
    )

    ann.fit(
        X_tr_ann,
        y_tr_ann,
        validation_data=(
            X_va_ann,
            y_scaler.transform(
                y_va.values.reshape(-1,1)
            )
        ),
        epochs=1000,
        batch_size=32,
        callbacks=[early_stop],
        verbose=0
    )

    ann_pred_scaled = ann.predict(
        X_va_ann,
        verbose=0
    )

    ann_pred = y_scaler.inverse_transform(
        ann_pred_scaled
    ).flatten()

    oof_ann[valid_idx] = ann_pred

    # ======================================================
    # METRICS
    # ======================================================

    print(
        "RF R2 =",
        round(
            r2_score(
                y_va,
                rf_pred
            ),
            6
        )
    )

    print(
        "XGB R2 =",
        round(
            r2_score(
                y_va,
                xgb_pred
            ),
            6
        )
    )

    print(
        "ANN R2 =",
        round(
            r2_score(
                y_va,
                ann_pred
            ),
            6
        )
    )
```

```python
# ==========================================================
# CELL 8: OOF RESULTS
# ==========================================================

results = pd.DataFrame({

    "Model": [
        "RF_OOF",
        "XGB_OOF",
        "ANN_OOF"
    ],

    "R2": [
        r2_score(y_train, oof_rf),
        r2_score(y_train, oof_xgb),
        r2_score(y_train, oof_ann)
    ],

    "MAE": [
        mean_absolute_error(y_train, oof_rf),
        mean_absolute_error(y_train, oof_xgb),
        mean_absolute_error(y_train, oof_ann)
    ],

    "RMSE": [
        np.sqrt(mean_squared_error(y_train, oof_rf)),
        np.sqrt(mean_squared_error(y_train, oof_xgb)),
        np.sqrt(mean_squared_error(y_train, oof_ann))
    ]
})

print("\nOOF RESULTS\n")
print(results)
```

```python
# ==========================================================
# CELL 9: CREATE META DATASET
# ==========================================================

meta_X = pd.DataFrame({

    "RF": oof_rf,
    "XGB": oof_xgb,
    "ANN": oof_ann

})

meta_y = y_train.reset_index(
    drop=True
)

print(meta_X.head())
print(meta_X.shape)
```

```python
# ==========================================================
# TRAIN & SAVE BASE MODELS
# ==========================================================

from sklearn.preprocessing import StandardScaler
from tensorflow.keras.callbacks import EarlyStopping
import joblib

# ==========================================================
# RF
# ==========================================================

rf.fit(X, y)

# ==========================================================
# XGB
# ==========================================================

xgb.fit(X, y)

# ==========================================================
# ANN
# ==========================================================

X_scaler_final = StandardScaler()
y_scaler_final = StandardScaler()

X_full_scaled = X_scaler_final.fit_transform(X)

y_full_scaled = y_scaler_final.fit_transform(
    y.values.reshape(-1, 1)
)

early_stop = EarlyStopping(
    monitor="loss",
    patience=50,
    restore_best_weights=True,
    verbose=1
)

final_ann = build_ann()

final_ann.fit(
    X_full_scaled,
    y_full_scaled,
    epochs=700,
    batch_size=32,
    callbacks=[early_stop],
    verbose=0
)

# ==========================================================
# SAVE RF
# ==========================================================

joblib.dump(
    rf,
    "SPGR91_RF_base.pkl"
)

# ==========================================================
# SAVE XGB
# ==========================================================

joblib.dump(
    xgb,
    "SPGR91_XGB_base.pkl"
)

# ==========================================================
# SAVE ANN
# ==========================================================

final_ann.save(
    "SPGR91_ANN_base.h5"
)

# ==========================================================
# SAVE SCALERS
# ==========================================================

joblib.dump(
    X_scaler_final,
    "SPGR91_ANN_X_scaler.pkl"
)

joblib.dump(
    y_scaler_final,
    "SPGR91_ANN_y_scaler.pkl"
)

print("\nBase Models Saved Successfully")
```

```python
# ==========================================================
# CELL 1: IMPORT LIBRARIES
# ==========================================================

import numpy as np
import pandas as pd
import joblib

from tensorflow.keras.models import load_model

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)
SP_GR = 0.76  # SAMPLE11
# ==========================================================
# CELL 2: LOAD TRAINED MODELS
# ==========================================================

rf_model = joblib.load(
    "SPGR91_RF_base.pkl"
)

xgb_model = joblib.load(
    "SPGR91_XGB_base.pkl"
)

ann_model = load_model(
    "SPGR91_ANN_base.h5"
)

ANN_X_scaler = joblib.load(
    "SPGR91_ANN_X_scaler.pkl"
)

ANN_y_scaler = joblib.load(
    "SPGR91_ANN_y_scaler.pkl"
)

linear_meta = joblib.load(
    "SPGR91_Linear_meta_WITHOUT RF.pkl"
)

xgb_meta = joblib.load(
    "SPGR91_XGB_meta_WITHOUT RF.pkl"
)

# ==========================================================
# SAMPLE 11 EXPERIMENTAL
# ==========================================================

pressure = np.array([
    590.00,
    702.22,
    814.44,
    926.67,
    1038.89,
    1151.11,
    1263.33,
    1375.56,
    1487.78,
    1600.00
])


# ==========================================================
# EXPERIMENTAL
# ==========================================================

experimental = np.array([
    287.94,
    289.30,
    290.32,
    291.09,
    291.66,
    292.12,
    292.54,
    292.98,
    293.53,
    294.26
])

# ==========================================================
# CELL 5: CREATE INPUT DATASET
# ==========================================================

X_new = pd.DataFrame({

    "Specific Gravity":[
        SP_GR
    ] * len(pressure),

    "Pressure (psia)":pressure

})

# ==========================================================
# CELL 6: BASE MODEL PREDICTIONS
# ==========================================================

rf_pred = rf_model.predict(
    X_new
)

xgb_pred = xgb_model.predict(
    X_new
)

# ==========================================================
# ANN
# ==========================================================

X_ann = ANN_X_scaler.transform(
    X_new
)

ann_pred_scaled = ann_model.predict(
    X_ann,
    verbose=0
)

ann_pred = ANN_y_scaler.inverse_transform(
    ann_pred_scaled
).flatten()

# ==========================================================
# META FEATURES
# ==========================================================

meta_features = pd.DataFrame({

    #"RF": rf_pred,

    "XGB": xgb_pred,

    "ANN": ann_pred

})

# ==========================================================
# HYBRID MODELS
# ==========================================================

hybrid_linear_pred = linear_meta.predict(
    meta_features
)

hybrid_xgb_pred = xgb_meta.predict(
    meta_features
)

# ==========================================================
# CELL 7: METRICS FUNCTION
# ==========================================================

def calculate_metrics(y_true, y_pred):

    r2 = r2_score(
        y_true,
        y_pred
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    rrmse = (
        rmse /
        np.mean(y_true)
    ) * 100

    return r2, mae, rmse, rrmse

# ==========================================================
# CELL 8: EXTERNAL VALIDATION
# ==========================================================

rf_metrics = calculate_metrics(
    experimental,
    rf_pred
)

xgb_metrics = calculate_metrics(
    experimental,
    xgb_pred
)

ann_metrics = calculate_metrics(
    experimental,
    ann_pred
)

linear_metrics = calculate_metrics(
    experimental,
    hybrid_linear_pred
)

xgbmeta_metrics = calculate_metrics(
    experimental,
    hybrid_xgb_pred
)

metrics = pd.DataFrame({

    "Model":[
        "RF",
        "XGB",
        "ANN",
        "Hybrid_Linear_Meta",
        "Hybrid_XGB_Meta"
    ],

    "R2":[
        rf_metrics[0],
        xgb_metrics[0],
        ann_metrics[0],
        linear_metrics[0],
        xgbmeta_metrics[0]
    ],

    "MAE":[
        rf_metrics[1],
        xgb_metrics[1],
        ann_metrics[1],
        linear_metrics[1],
        xgbmeta_metrics[1]
    ],

    "RMSE":[
        rf_metrics[2],
        xgb_metrics[2],
        ann_metrics[2],
        linear_metrics[2],
        xgbmeta_metrics[2]
    ],

    "RRMSE (%)":[
        rf_metrics[3],
        xgb_metrics[3],
        ann_metrics[3],
        linear_metrics[3],
        xgbmeta_metrics[3]
    ]

})

print("\nExternal Validation Results\n")

print(
    metrics.sort_values(
        by="R2",
        ascending=False
    )
)

# ==========================================================
# CELL 9: PREDICTIONS TABLE
# ==========================================================

results = pd.DataFrame({

    "Pressure (psia)":pressure,

    "RF_Pred":rf_pred,

    "XGB_Pred":xgb_pred,

    "ANN_Pred":ann_pred,

    "Hybrid_Linear_Pred":hybrid_linear_pred,

    "Hybrid_XGB_Pred":hybrid_xgb_pred

})

print(results)

results.to_excel(
    "SPGR_Predictions.xlsx",
    index=False
)
```

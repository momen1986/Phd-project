```python
# ==========================================================
# CELL 1: IMPORT REQUIRED LIBRARIES
# ==========================================================
# This section imports all required libraries for:
# 1. Data manipulation and numerical operations.
# 2. Machine learning model development.
# 3. GroupKFold cross-validation.
# 4. Stacking ensemble construction.
# 5. Model evaluation and saving.
# ==========================================================

import numpy as np
import pandas as pd
import joblib
import tensorflow as tf
import random

from sklearn.model_selection import GroupKFold

from sklearn.ensemble import RandomForestRegressor

from sklearn.preprocessing import StandardScaler

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

from xgboost import XGBRegressor

from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from tensorflow.keras.optimizers import Adam
import joblib
import tensorflow as tf
import random

SEED = 42

np.random.seed(SEED)

random.seed(SEED)

tf.random.set_seed(SEED)
# ==========================================================
# CELL 2: LOAD COMPOSITION DATASET
# ==========================================================
# This section loads the composition dataset and separates:
# - Input features (X)
# - Target variable (Hydrate Temperature)
# - Group labels used for GroupKFold validation
#
# Group labels are retained to prevent data leakage
# during the generation of Out-of-Fold predictions.
# ==========================================================

df = pd.read_excel("composition_dataset_clean.xlsx")

TARGET = "Hydrate Temperature (K)"

feature_cols = [
    c for c in df.columns
    if c != TARGET
]

group_cols = [
    c for c in feature_cols
    if c not in ["Pressure (psia)", "PR"]
]

df["group_id"] = (
    df[group_cols]
    .round(3)
    .astype(str)
    .agg("_".join, axis=1)
)

X = df.drop(columns=[TARGET, "group_id"])

y = df[TARGET]

groups = df["group_id"]

# ==========================================================
# CELL 3: DEFINE BASE LEARNERS
# ==========================================================
# This section initializes the optimized machine learning
# models using the best hyperparameters obtained during
# previous model development and optimization.
#
# Base learners:
# - Random Forest (RF)
# - Extreme Gradient Boosting (XGB)
# - Artificial Neural Network (ANN)
# ==========================================================

rf = RandomForestRegressor(
    n_estimators=300,
    max_depth=15,
    min_samples_split=2,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1
)

xgb = XGBRegressor(
    n_estimators=500,
    learning_rate=0.03,
    max_depth=4,
    subsample=0.8,
    colsample_bytree=0.8,
    random_state=42,
    objective='reg:squarederror'
)

# ==========================================================
# ANN ARCHITECTURE
# ==========================================================

def build_ann():

    model = Sequential()

    model.add(
        Dense(
            128,
            activation='relu',
            input_shape=(X.shape[1],)
        )
    )

    model.add(Dense(64, activation='relu'))
    model.add(Dense(32, activation='relu'))
    model.add(Dense(16, activation='relu'))

    model.add(
        Dense(
            1,
            activation='linear'
        )
    )

    model.compile(
        optimizer=Adam(
            learning_rate=0.0005
        ),
        loss='mse'
    )

    return model

# ==========================================================
# CELL 4: INITIALIZE GROUPKFOLD AND OOF ARRAYS
# ==========================================================
# Out-of-Fold (OOF) predictions are generated to eliminate
# data leakage during stacking.
#
# Each prediction is produced from a model that has not
# previously seen the corresponding validation samples.
#
# These OOF predictions will later serve as inputs for
# training the meta-learner.
# ==========================================================

gkf = GroupKFold(n_splits=5)

oof_rf = np.zeros(len(X))
oof_xgb = np.zeros(len(X))
oof_ann = np.zeros(len(X))

# ==========================================================
# CELL 5: GENERATE OOF PREDICTIONS
# ==========================================================
# For each GroupKFold split:
# 1. Train RF and XGB.
# 2. Scale X and y for ANN.
# 3. Train ANN on scaled targets.
# 4. Convert ANN predictions back to original units.
# 5. Store OOF predictions.
# ==========================================================

# ==========================================================
# GENERATE OOF PREDICTIONS + FOLD METRICS
# ==========================================================

from tensorflow.keras.callbacks import EarlyStopping

for fold, (train_idx, valid_idx) in enumerate(
        gkf.split(X, y, groups)
):

    print("\n" + "="*60)
    print(f"FOLD {fold+1}")
    print("="*60)

    # ======================================================
    # SPLIT
    # ======================================================

    X_train = X.iloc[train_idx]
    X_valid = X.iloc[valid_idx]

    y_train = y.iloc[train_idx]
    y_valid = y.iloc[valid_idx]

    # ======================================================
    # RF
    # ======================================================

    rf.fit(
        X_train,
        y_train
    )

    rf_pred = rf.predict(
        X_valid
    )

    oof_rf[valid_idx] = rf_pred

    rf_r2 = r2_score(
        y_valid,
        rf_pred
    )

    # ======================================================
    # XGB
    # ======================================================

    xgb.fit(
        X_train,
        y_train
    )

    xgb_pred = xgb.predict(
        X_valid
    )

    oof_xgb[valid_idx] = xgb_pred

    xgb_r2 = r2_score(
        y_valid,
        xgb_pred
    )

    # ======================================================
    # ANN SCALING
    # ======================================================

    X_scaler = StandardScaler()

    y_scaler = StandardScaler()

    X_train_ann = X_scaler.fit_transform(
        X_train
    )

    X_valid_ann = X_scaler.transform(
        X_valid
    )

    y_train_ann = y_scaler.fit_transform(
        y_train.values.reshape(-1, 1)
    )

    y_valid_ann = y_scaler.transform(
        y_valid.values.reshape(-1, 1)
    )

    # ======================================================
    # ANN
    # ======================================================

    ann = build_ann()

    early_stop = EarlyStopping(
        monitor="val_loss",
        patience=50,
        restore_best_weights=True,
        verbose=0
    )

    ann.fit(
        X_train_ann,
        y_train_ann,
        validation_data=(
            X_valid_ann,
            y_valid_ann
        ),
        epochs=1000,
        batch_size=32,
        callbacks=[early_stop],
        verbose=0
    )

    ann_pred_scaled = ann.predict(
        X_valid_ann,
        verbose=0
    )

    ann_pred = y_scaler.inverse_transform(
        ann_pred_scaled
    ).flatten()

    oof_ann[valid_idx] = ann_pred

    ann_r2 = r2_score(
        y_valid,
        ann_pred
    )

    ann_rmse = np.sqrt(
        mean_squared_error(
            y_valid,
            ann_pred
        )
    )

    # ======================================================
    # PRINT RESULTS
    # ======================================================

    print(
        f"RF  R² = {rf_r2:.6f}"
    )

    print(
        f"XGB R² = {xgb_r2:.6f}"
    )

    print(
        f"ANN R² = {ann_r2:.6f}"
    )

    print(
        f"ANN RMSE = {ann_rmse:.6f}"
    )

    print(
        f"Train Samples = {len(train_idx)}"
    )

    print(
        f"Valid Samples = {len(valid_idx)}"
    )
```

```python
# ==========================================================
# CELL 6: OOF PERFORMANCE OF BASE LEARNERS
# ==========================================================

def calculate_metrics(y_true, y_pred):

    r2 = r2_score(y_true, y_pred)

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    rrmse = (
        rmse /
        np.mean(y_true)
    ) * 100

    return r2, mae, rmse, rrmse


rf_metrics = calculate_metrics(
    y,
    oof_rf
)

xgb_metrics = calculate_metrics(
    y,
    oof_xgb
)

ann_metrics = calculate_metrics(
    y,
    oof_ann
)

oof_results = pd.DataFrame({

    "Model":[
        "RF_OOF",
        "XGB_OOF",
        "ANN_OOF"
    ],

    "R2":[
        rf_metrics[0],
        xgb_metrics[0],
        ann_metrics[0]
    ],

    "MAE":[
        rf_metrics[1],
        xgb_metrics[1],
        ann_metrics[1]
    ],

    "RMSE":[
        rf_metrics[2],
        xgb_metrics[2],
        ann_metrics[2]
    ],

    "RRMSE (%)":[
        rf_metrics[3],
        xgb_metrics[3],
        ann_metrics[3]
    ]

})

print("\nOOF RESULTS\n")

print(oof_results)
```

```python
# ==========================================================
# TRAIN BASE MODELS
# RF + XGB + ANN
# ==========================================================

from sklearn.preprocessing import StandardScaler
import joblib

# ==========================================================
# RETRAIN RF ON FULL DATA
# ==========================================================

rf.fit(
    X,
    y
)

# ==========================================================
# RETRAIN XGB ON FULL DATA
# ==========================================================

xgb.fit(
    X,
    y
)

# ==========================================================
# RETRAIN ANN ON FULL DATA
# ==========================================================

final_X_scaler = StandardScaler()
final_y_scaler = StandardScaler()

X_scaled = final_X_scaler.fit_transform(
    X
)

y_scaled = final_y_scaler.fit_transform(
    y.values.reshape(-1, 1)
)

final_ann = build_ann()

final_ann.fit(
    X_scaled,
    y_scaled,
    epochs=700,
    batch_size=32,
    verbose=0
)

print("Base Models Trained Successfully")

# ==========================================================
# SAVE BASE LEARNERS
# ==========================================================

joblib.dump(
    rf,
    "RF888_base.pkl"
)

joblib.dump(
    xgb,
    "XGB888_base.pkl"
)

final_ann.save(
    "ANN888_base.h5"
)

joblib.dump(
    final_X_scaler,
    "ANN888_X_scaler.pkl"
)

joblib.dump(
    final_y_scaler,
    "ANN888_y_scaler.pkl"
)

print("Base Models Saved Successfully")
```

```python
# ==========================================================
# TRAIN META MODELS
# RF + XGB + ANN STACKING
# ==========================================================

from sklearn.linear_model import LinearRegression
from xgboost import XGBRegressor
import pandas as pd
import joblib

# ==========================================================
# CREATE META DATASET
# ==========================================================

meta_X = pd.DataFrame({

    #"RF": oof_rf,

    "XGB": oof_xgb,

    "ANN": oof_ann

})

meta_y = y.copy()

print("Meta X Shape :", meta_X.shape)
print("Meta y Shape :", meta_y.shape)

# ==========================================================
# TRAIN XGB META
# ==========================================================

xgb_meta = XGBRegressor(
    n_estimators=600,
    max_depth=4,
    learning_rate=0.05,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

xgb_meta.fit(
    meta_X,
    meta_y
)

print("XGB Meta Trained")

# ==========================================================
# TRAIN LINEAR META
# ==========================================================

linear_meta = LinearRegression()

linear_meta.fit(
    meta_X,
    meta_y
)

print("Linear Meta Trained")

# ==========================================================
# SAVE META LEARNERS
# ==========================================================

joblib.dump(
    xgb_meta,
    "XGB888_meta_WITHOUT RF.pkl"
)

joblib.dump(
    linear_meta,
    "Linear888_meta_WITHOUT RF.pkl"
)

print("Meta Models Saved Successfully")
```

```python
# ==========================================================
# CELL 1: IMPORT REQUIRED LIBRARIES
# ==========================================================
# This section imports all libraries required for:
# 1. Loading trained models.
# 2. External validation.
# 3. Generating predictions.
# 4. Calculating performance metrics.
# ==========================================================

import numpy as np
import pandas as pd
import joblib

from tensorflow.keras.models import load_model

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

# ==========================================================
# CELL 2: LOAD TRAINED MODELS
# ==========================================================
# Load the trained base learners, ANN scalers,
# and the stacking meta-learner.
# ==========================================================

rf_model = joblib.load(
    "RF888_base.pkl"
)

xgb_model = joblib.load(
    "XGB888_base.pkl"
)

ann_model = load_model(
    "ANN888_base.h5"
)

ANN_X_scaler = joblib.load(
    "ANN888_X_scaler.pkl"
)

ANN_y_scaler = joblib.load(
    "ANN888_y_scaler.pkl"
)

# ==========================================================
# LOAD META MODELS
# ==========================================================

xgb_meta = joblib.load(
    "XGB888_meta_WITHOUT RF.pkl"
)

linear_meta = joblib.load(
    "Linear888_meta_WITHOUT RF.pkl"
)



# ==========================================================
# GAS COMPOSITION (SALMAN)
# ==========================================================

CH4 = 74.10
CO2 = 3.95
C2  = 10.82
C3  = 6.70
iC4 = 0.94
nC4 = 2.04
iC5 = 0.53
nC5 = 0.58
C6  = 0.25
C7  = 0.08
Water=0.0

# ==========================================================
# PRESSURE
# ==========================================================

pressure = np.array([
     700.00,
     891.67,
    1083.33,
    1275.00,
    1466.67,
    1658.33,
    1850.00,
    2041.67,
    2233.33,
    2425.00,
    2616.67,
    2808.33,
    3000.00
])

# ==========================================================
# EXPERIMENTAL
# ==========================================================

experimental = np.array([
    289.86,
    291.93,
    293.02,
    293.68,
    294.26,
    294.79,
    295.26,
    295.68,
    296.07,
    296.44,
    296.79,
    297.13,
    297.49
])

# ==========================================================
# CELL 5: CALCULATE SPECIFIC GRAVITY AND PR
# ==========================================================
# Calculate gas specific gravity and pseudo-reduced
# pressure using the same procedure adopted during
# model development.
# ==========================================================

mw_mix = (
      CH4*16.043
    + CO2*44.010
    + C2*30.070
    + C3*44.097
    + iC4*58.124
    + nC4*58.124
    + iC5*72.151
    + nC5*72.151
    + C6*86.178
    + C7*100.205
)/100

SP_GR = mw_mix / 28.97

PPC = (
      CH4*667.0
    + CO2*1071.0
    + C2*707.8
    + C3*616.0
    + iC4*529.1
    + nC4*550.7
    + iC5*490.4
    + nC5*488.6
    + C6*436.9
    + C7*397.0
)/100

PR = pressure / PPC

# ==========================================================
# CELL 6: CREATE INPUT DATASET
# ==========================================================
# Construct the feature matrix using the same feature
# order employed during model training.
# ==========================================================

X_new = pd.DataFrame({

    "Water":[Water]*len(pressure),

    "Methane":[CH4]*len(pressure),
    "Ethane":[C2]*len(pressure),
    "Propane":[C3]*len(pressure),
    "Isobutane":[iC4]*len(pressure),
    "Butane":[nC4]*len(pressure),
    "Isopentane":[iC5]*len(pressure),
    "Pentane":[nC5]*len(pressure),
    "Hexane":[C6]*len(pressure),
    "Heptane":[C7]*len(pressure),

    "Carbon Dioxide":[CO2]*len(pressure),

    "SPECIFIC GRAVITY":[SP_GR]*len(pressure),

    "Pressure (psia)":pressure,

    "PR":PR

})

# ==========================================================
# CELL 7: GENERATE BASE MODEL PREDICTIONS
# ==========================================================
# Generate predictions from RF, XGB and ANN models.
# ==========================================================

rf_pred = rf_model.predict(
    X_new
)

xgb_pred = xgb_model.predict(
    X_new
)

# ==========================================================
# ANN PREDICTION
# ==========================================================

X_ann = ANN_X_scaler.transform(
    X_new
)

ann_pred_scaled = ann_model.predict(
    X_ann,
    verbose=0
)

ann_pred = ANN_y_scaler.inverse_transform(
    ann_pred_scaled
).flatten()

# ==========================================================
# CELL 8: GENERATE HYBRID PREDICTIONS
# ==========================================================

meta_features = pd.DataFrame({

    #"RF": rf_pred,

    "XGB": xgb_pred,

    "ANN": ann_pred

})

# ==========================================================
# XGB META
# ==========================================================

hybrid_xgb_pred = xgb_meta.predict(
    meta_features
)

# ==========================================================
# LINEAR META
# ==========================================================

hybrid_linear_pred = linear_meta.predict(
    meta_features
)

# ==========================================================
# CELL 9: DEFINE EVALUATION METRICS
# ==========================================================
# Calculate R², MAE, RMSE, and RRMSE for model
# performance assessment.
# ==========================================================

def calculate_metrics(y_true, y_pred):

    r2 = r2_score(
        y_true,
        y_pred
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    rrmse = (
        rmse /
        np.mean(y_true)
    ) * 100

    return r2, mae, rmse, rrmse

# ==========================================================
# CELL 10: CALCULATE PERFORMANCE METRICS
# ==========================================================

rf_metrics = calculate_metrics(
    experimental,
    rf_pred
)

xgb_metrics = calculate_metrics(
    experimental,
    xgb_pred
)

ann_metrics = calculate_metrics(
    experimental,
    ann_pred
)

hybrid_xgb_metrics = calculate_metrics(
    experimental,
    hybrid_xgb_pred
)

hybrid_linear_metrics = calculate_metrics(
    experimental,
    hybrid_linear_pred
)
# ==========================================================
# CELL 11: CREATE METRICS TABLE
# ==========================================================

metrics = pd.DataFrame({

    "Model":[
        "RF",
        "XGB",
        "ANN",
        "Hybrid_XGB_Meta",
        "Hybrid_Linear_Meta"
    ],

    "R2":[
        rf_metrics[0],
        xgb_metrics[0],
        ann_metrics[0],
        hybrid_xgb_metrics[0],
        hybrid_linear_metrics[0]
    ],

    "MAE":[
        rf_metrics[1],
        xgb_metrics[1],
        ann_metrics[1],
        hybrid_xgb_metrics[1],
        hybrid_linear_metrics[1]
    ],

    "RMSE":[
        rf_metrics[2],
        xgb_metrics[2],
        ann_metrics[2],
        hybrid_xgb_metrics[2],
        hybrid_linear_metrics[2]
    ],

    "RRMSE (%)":[
        rf_metrics[3],
        xgb_metrics[3],
        ann_metrics[3],
        hybrid_xgb_metrics[3],
        hybrid_linear_metrics[3]
    ]

})

print("\nExternal Validation Results\n")

print(
    metrics.sort_values(
        by="R2",
        ascending=False
    )
)
```

```python
# ==========================================================
# CELL 1: IMPORT REQUIRED LIBRARIES
# ==========================================================
# This section imports all libraries required for:
# 1. Loading trained models.
# 2. External validation.
# 3. Generating predictions.
# 4. Calculating performance metrics.
# ==========================================================

import numpy as np
import pandas as pd
import joblib

from tensorflow.keras.models import load_model

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

# ==========================================================
# CELL 2: LOAD TRAINED MODELS
# ==========================================================
# Load the trained base learners, ANN scalers,
# and the stacking meta-learner.
# ==========================================================

rf_model = joblib.load(
    "RF888_base.pkl"
)

xgb_model = joblib.load(
    "XGB888_base.pkl"
)

ann_model = load_model(
    "ANN888_base.h5"
)

ANN_X_scaler = joblib.load(
    "ANN888_X_scaler.pkl"
)

ANN_y_scaler = joblib.load(
    "ANN888_y_scaler.pkl"
)

# ==========================================================
# LOAD META MODELS
# ==========================================================

xgb_meta = joblib.load(
    "XGB888_meta_WITHOUT RF.pkl"
)

linear_meta = joblib.load(
    "Linear888_meta_WITHOUT RF.pkl"
)

# ==========================================================
# SAMPLE 11 COMPOSITION
# ==========================================================

CH4  = 77.61
CO2  = 6.14
C2   = 8.81
C3   = 3.37
iC4  = 0.75
nC4  = 1.33
iC5  = 0.44
nC5  = 0.69
C6   = 0.86
C7   = 0.00

Water = 0.0
N2    = 0.0
H2S   = 0.0

# ==========================================================
# SAMPLE 11 EXPERIMENTAL
# ==========================================================

pressure = np.array([
    590.00,
    702.22,
    814.44,
    926.67,
    1038.89,
    1151.11,
    1263.33,
    1375.56,
    1487.78,
    1600.00
])


# ==========================================================
# EXPERIMENTAL
# ==========================================================

experimental = np.array([
    287.94,
    289.30,
    290.32,
    291.09,
    291.66,
    292.12,
    292.54,
    292.98,
    293.53,
    294.26
])

# ==========================================================
# CELL 5: CALCULATE SPECIFIC GRAVITY AND PR
# ==========================================================
# Calculate gas specific gravity and pseudo-reduced
# pressure using the same procedure adopted during
# model development.
# ==========================================================

mw_mix = (
      CH4*16.043
    + CO2*44.010
    + C2*30.070
    + C3*44.097
    + iC4*58.124
    + nC4*58.124
    + iC5*72.151
    + nC5*72.151
    + C6*86.178
    + C7*100.205
)/100

SP_GR = mw_mix / 28.97

PPC = (
      CH4*667.0
    + CO2*1071.0
    + C2*707.8
    + C3*616.0
    + iC4*529.1
    + nC4*550.7
    + iC5*490.4
    + nC5*488.6
    + C6*436.9
    + C7*397.0
)/100

PR = pressure / PPC

# ==========================================================
# CELL 6: CREATE INPUT DATASET
# ==========================================================
# Construct the feature matrix using the same feature
# order employed during model training.
# ==========================================================

X_new = pd.DataFrame({

    "Water":[Water]*len(pressure),

    "Methane":[CH4]*len(pressure),
    "Ethane":[C2]*len(pressure),
    "Propane":[C3]*len(pressure),
    "Isobutane":[iC4]*len(pressure),
    "Butane":[nC4]*len(pressure),
    "Isopentane":[iC5]*len(pressure),
    "Pentane":[nC5]*len(pressure),
    "Hexane":[C6]*len(pressure),
    "Heptane":[C7]*len(pressure),

    "Carbon Dioxide":[CO2]*len(pressure),

    "SPECIFIC GRAVITY":[SP_GR]*len(pressure),

    "Pressure (psia)":pressure,

    "PR":PR

})

# ==========================================================
# CELL 7: GENERATE BASE MODEL PREDICTIONS
# ==========================================================
# Generate predictions from RF, XGB and ANN models.
# ==========================================================

rf_pred = rf_model.predict(
    X_new
)

xgb_pred = xgb_model.predict(
    X_new
)

# ==========================================================
# ANN PREDICTION
# ==========================================================

X_ann = ANN_X_scaler.transform(
    X_new
)

ann_pred_scaled = ann_model.predict(
    X_ann,
    verbose=0
)

ann_pred = ANN_y_scaler.inverse_transform(
    ann_pred_scaled
).flatten()

# ==========================================================
# CELL 8: GENERATE HYBRID PREDICTIONS
# ==========================================================

meta_features = pd.DataFrame({

    #"RF": rf_pred,

    "XGB": xgb_pred,

    "ANN": ann_pred

})

# ==========================================================
# XGB META
# ==========================================================

hybrid_xgb_pred = xgb_meta.predict(
    meta_features
)

# ==========================================================
# LINEAR META
# ==========================================================

hybrid_linear_pred = linear_meta.predict(
    meta_features
)

# ==========================================================
# CELL 9: DEFINE EVALUATION METRICS
# ==========================================================
# Calculate R², MAE, RMSE, and RRMSE for model
# performance assessment.
# ==========================================================

def calculate_metrics(y_true, y_pred):

    r2 = r2_score(
        y_true,
        y_pred
    )

    mae = mean_absolute_error(
        y_true,
        y_pred
    )

    rmse = np.sqrt(
        mean_squared_error(
            y_true,
            y_pred
        )
    )

    rrmse = (
        rmse /
        np.mean(y_true)
    ) * 100

    return r2, mae, rmse, rrmse

# ==========================================================
# CELL 10: CALCULATE PERFORMANCE METRICS
# ==========================================================

rf_metrics = calculate_metrics(
    experimental,
    rf_pred
)

xgb_metrics = calculate_metrics(
    experimental,
    xgb_pred
)

ann_metrics = calculate_metrics(
    experimental,
    ann_pred
)

hybrid_xgb_metrics = calculate_metrics(
    experimental,
    hybrid_xgb_pred
)

hybrid_linear_metrics = calculate_metrics(
    experimental,
    hybrid_linear_pred
)
# ==========================================================
# SHOW ALL PREDICTIONS
# ==========================================================

predictions_df = pd.DataFrame({

    "Pressure (psia)": pressure,

    "Experimental": experimental,

    "RF_Pred": rf_pred,

    "XGB_Pred": xgb_pred,

    "ANN_Pred": ann_pred,

    "Hybrid_XGB_Pred": hybrid_xgb_pred,

    "Hybrid_Linear_Pred": hybrid_linear_pred

})

pd.set_option(
    'display.max_columns',
    None
)

pd.set_option(
    'display.width',
    None
)

print(predictions_df.round(4))
# ==========================================================
# CELL 11: CREATE METRICS TABLE
# ==========================================================

metrics = pd.DataFrame({

    "Model":[
        "RF",
        "XGB",
        "ANN",
        "Hybrid_XGB_Meta",
        "Hybrid_Linear_Meta"
    ],

    "R2":[
        rf_metrics[0],
        xgb_metrics[0],
        ann_metrics[0],
        hybrid_xgb_metrics[0],
        hybrid_linear_metrics[0]
    ],

    "MAE":[
        rf_metrics[1],
        xgb_metrics[1],
        ann_metrics[1],
        hybrid_xgb_metrics[1],
        hybrid_linear_metrics[1]
    ],

    "RMSE":[
        rf_metrics[2],
        xgb_metrics[2],
        ann_metrics[2],
        hybrid_xgb_metrics[2],
        hybrid_linear_metrics[2]
    ],

    "RRMSE (%)":[
        rf_metrics[3],
        xgb_metrics[3],
        ann_metrics[3],
        hybrid_xgb_metrics[3],
        hybrid_linear_metrics[3]
    ]

})

print("\nExternal Validation Results\n")

print(
    metrics.sort_values(
        by="R2",
        ascending=False
    )
)
```

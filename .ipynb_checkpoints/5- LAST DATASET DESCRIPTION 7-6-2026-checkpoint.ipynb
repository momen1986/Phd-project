# 1- CREATING UNORDED DATASET 

```python
import pandas as pd
from collections import deque

df = pd.read_excel("composition_dataset_clean.xlsx")

composition_cols = [
    "Water","Methane","Ethane","Propane",
    "Isobutane","Butane","Isopentane",
    "Pentane","Hexane","Heptane",
    "Carbon Dioxide"
]

# إنشاء معرف للتركيب
df["Comp_ID"] = (
    df[composition_cols]
    .round(6)
    .astype(str)
    .agg("_".join, axis=1)
)

# تقسيم الداتا حسب التركيب
groups = [
    g.reset_index(drop=True)
    for _, g in df.groupby("Comp_ID")
]

# ترتيب المجموعات من الأكبر للأصغر
groups = deque(
    sorted(groups, key=len, reverse=True)
)

rows = []

# سحب صف واحد من كل مجموعة بالتناوب
while groups:

    next_round = deque()

    while groups:

        g = groups.popleft()

        rows.append(g.iloc[[0]])

        if len(g) > 1:
            next_round.append(
                g.iloc[1:].reset_index(drop=True)
            )

    groups = next_round

# دمج الصفوف
df_display = pd.concat(
    rows,
    ignore_index=True
)

df_display.to_excel(
    "Composition_Dataset_Display.xlsx",
    index=False
)

print(df_display.head(20))
```

```python
import pandas as pd
from collections import deque

df = pd.read_csv("sp_gr_dataset_clean.csv")

# تقسيم SG إلى مجموعات
df["SG_Group"] = pd.qcut(
    df["Specific Gravity"],
    q=20,          # يمكن تغييرها إلى 10 أو 15 أو 25
    duplicates="drop"
)

# إنشاء المجموعات
groups = [
    g.reset_index(drop=True)
    for _, g in df.groupby("SG_Group")
]

# ترتيبها من الأكبر للأصغر
groups = deque(
    sorted(groups, key=len, reverse=True)
)

rows = []

# سحب صف واحد من كل مجموعة بالتناوب
while groups:

    next_round = deque()

    while groups:

        g = groups.popleft()

        rows.append(g.iloc[[0]])

        if len(g) > 1:
            next_round.append(
                g.iloc[1:].reset_index(drop=True)
            )

    groups = next_round

df_display = pd.concat(
    rows,
    ignore_index=True
)

df_display.drop(
    columns=["SG_Group"],
    inplace=True
)

df_display.to_excel(
    "SP_GR_Dataset_Display.xlsx",
    index=False
)

print(df_display.head(20))
```

# 2- CREATING DISPERSION & descrption and effect of each column (MI) on TH FOR THE COMPOSITION DATASET

```python
import pandas as pd
import numpy as np

# =====================================================
# LOAD DATA
# =====================================================

df = pd.read_excel("composition_dataset_clean.xlsx")

# =====================================================
# COMPOSITION COLUMNS
# =====================================================

composition_cols = [
    "Water",
    "Methane",
    "Ethane",
    "Propane",
    "Isobutane",
    "Butane",
    "Isopentane",
    "Pentane",
    "Hexane",
    "Heptane",
    "Carbon Dioxide"
]

# =====================================================
# RANDOM SEED
# =====================================================

np.random.seed(42)

# =====================================================
# SMALL DISPERSION
# =====================================================
# 0.05 mol% standard deviation
# يمكنك زيادة القيمة إلى 0.1 إذا أردت

noise = np.random.normal(
    loc=0,
    scale=0.05,
    size=df[composition_cols].shape
)

df_disp = df.copy()

df_disp[composition_cols] = (
    df_disp[composition_cols] + noise
)

# منع القيم السالبة

df_disp[composition_cols] = (
    df_disp[composition_cols]
    .clip(lower=0)
)

# =====================================================
# RENORMALIZE TO 100%
# =====================================================

row_sum = df_disp[composition_cols].sum(axis=1)

df_disp[composition_cols] = (
    df_disp[composition_cols]
    .div(row_sum, axis=0)
    * 100
)

# =====================================================
# SAVE
# =====================================================

df_disp.to_csv(
    "composition_dataset_dispersion.csv",
    index=False
)

print("Saved Successfully")
print(df_disp.head())
```

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# =====================================================
# LOAD DATA
# =====================================================

df = pd.read_csv("composition_dataset_dispersion.csv")

# =====================================================
# BASIC INFORMATION
# =====================================================

print("\n" + "="*70)
print("DATASET SHAPE")
print("="*70)
print(df.shape)

print("\n" + "="*70)
print("DATA TYPES")
print("="*70)
print(df.dtypes)

# =====================================================
# MISSING VALUES
# =====================================================

print("\n" + "="*70)
print("MISSING VALUES")
print("="*70)

missing = pd.DataFrame({
    "Missing Count": df.isnull().sum(),
    "Missing %": (df.isnull().sum()/len(df))*100
})

print(missing)

# =====================================================
# UNIQUE VALUES
# =====================================================

print("\n" + "="*70)
print("UNIQUE VALUES")
print("="*70)

unique_df = pd.DataFrame({
    "Unique Values": df.nunique()
})

print(unique_df)

# =====================================================
# DUPLICATE ROWS
# =====================================================

print("\n" + "="*70)
print("DUPLICATE ROWS")
print("="*70)

print(df.duplicated().sum())

# =====================================================
# ZERO VALUES
# =====================================================

print("\n" + "="*70)
print("ZERO VALUES")
print("="*70)

zero_df = pd.DataFrame({
    "Zero Count": (df == 0).sum()
})

print(zero_df)

# =====================================================
# DESCRIPTIVE STATISTICS
# =====================================================

print("\n" + "="*70)
print("DESCRIPTIVE STATISTICS")
print("="*70)

print(df.describe().T)

# =====================================================
# NUMERIC COLUMNS
# =====================================================

numeric_cols = df.select_dtypes(include=np.number).columns

# =====================================================
# CORRELATION MATRIX
# =====================================================

corr = df.corr(numeric_only=True)

print("\n" + "="*70)
print("CORRELATION MATRIX")
print("="*70)

print(corr)

# =====================================================
# FIGURE 1: MISSING VALUES
# =====================================================

plt.figure(figsize=(12,5))

missing["Missing Count"].plot(kind="bar")

plt.title("Missing Values per Feature")
plt.ylabel("Count")
plt.xlabel("Features")

plt.tight_layout()
plt.show()

# =====================================================
# FIGURE 2: UNIQUE VALUES
# =====================================================

plt.figure(figsize=(12,5))

unique_df["Unique Values"].sort_values().plot(kind="bar")

plt.title("Unique Values per Feature")
plt.ylabel("Count")
plt.xlabel("Features")

plt.tight_layout()
plt.show()

# =====================================================
# FIGURE 3: CORRELATION HEATMAP
# =====================================================

plt.figure(figsize=(12,10))

im = plt.imshow(corr)

plt.colorbar(im)

plt.xticks(
    range(len(corr.columns)),
    corr.columns,
    rotation=90
)

plt.yticks(
    range(len(corr.columns)),
    corr.columns
)

plt.title("Correlation Matrix")

plt.tight_layout()
plt.show()

# =====================================================
# FIGURE 4: HISTOGRAMS
# =====================================================

df[numeric_cols].hist(
    figsize=(20,15),
    bins=30
)

plt.suptitle("Feature Distributions")
plt.tight_layout()

plt.show()

# =====================================================
# FIGURE 5: BOXPLOTS
# =====================================================

plt.figure(figsize=(16,6))

df[numeric_cols].boxplot(
    rot=90
)

plt.title("Boxplots of Numerical Features")

plt.tight_layout()
plt.show()

# =====================================================
# FIGURE 6: TARGET DISTRIBUTION
# =====================================================

target_candidates = [
    "Hydrate Temperature (K)",
    "Hydrate Temperature",
    "Temperature (K)"
]

target = None

for col in target_candidates:
    if col in df.columns:
        target = col
        break

if target:

    plt.figure(figsize=(8,5))

    plt.hist(
        df[target],
        bins=30
    )

    plt.title(f"Distribution of {target}")
    plt.xlabel(target)
    plt.ylabel("Frequency")

    plt.tight_layout()
    plt.show()

# =====================================================
# FIGURE 7: PRESSURE VS TARGET
# =====================================================

pressure_candidates = [
    "Pressure (psia)",
    "Pressure"
]

pressure = None

for col in pressure_candidates:
    if col in df.columns:
        pressure = col
        break

if pressure and target:

    plt.figure(figsize=(8,6))

    plt.scatter(
        df[pressure],
        df[target],
        alpha=0.5
    )

    plt.xlabel(pressure)
    plt.ylabel(target)

    plt.title(f"{pressure} vs {target}")

    plt.tight_layout()
    plt.show()

# =====================================================
# FIGURE 8: FEATURE CORRELATION WITH TARGET
# =====================================================

if target:

    corr_target = (
        df.corr(numeric_only=True)[target]
        .sort_values(ascending=False)
    )

    plt.figure(figsize=(10,5))

    corr_target.plot(kind="bar")

    plt.title(f"Correlation with {target}")
    plt.ylabel("Correlation Coefficient")

    plt.tight_layout()
    plt.show()
```

```python
target = "Hydrate Temperature (K)"

corr_target = (
    df.corr(numeric_only=True)[target]
    .sort_values(ascending=False)
)

print(corr_target)
```

```python
from sklearn.feature_selection import mutual_info_regression
import pandas as pd

target = "Hydrate Temperature (K)"

X = df.drop(columns=[target])
y = df[target]

mi = mutual_info_regression(
    X,
    y,
    random_state=42
)

mi_df = pd.DataFrame({
    "Feature": X.columns,
    "MI": mi
})

print(
    mi_df.sort_values(
        "MI",
        ascending=False
    )
)
```

# 3- CREATING SP_GR DATA SET DISPERSION & MI DECRIPTION 

```python
import pandas as pd
import numpy as np

# Load data
df = pd.read_csv("sp_gr_dataset_clean.csv")

# Copy
df_disp = df.copy()

# Small dispersion on Specific Gravity
np.random.seed(42)

df_disp["Specific Gravity"] += np.random.normal(
    loc=0,
    scale=0.002,
    size=len(df_disp)
)

# Keep values realistic
df_disp["Specific Gravity"] = (
    df_disp["Specific Gravity"]
    .clip(lower=0.55, upper=1.20)
)

# Recalculate LOG P if exists
if "LOG P" in df_disp.columns:
    df_disp["LOG P"] = np.log(
        df_disp["Pressure (psia)"]
    )

# Save CSV
df_disp.to_csv(
    "sp_gr_dataset_dispersion.csv",
    index=False
)

print("Done")
print(df_disp.head())
```

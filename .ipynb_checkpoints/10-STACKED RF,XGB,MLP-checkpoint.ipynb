```python
# ==========================================================
# CELL 1: IMPORT REQUIRED LIBRARIES
# ==========================================================
# SP_GR HYBRID STACKING
# RF + XGB + ANN
# ==========================================================

import numpy as np
import pandas as pd
import random
import joblib
import tensorflow as tf

from sklearn.model_selection import GroupKFold

from sklearn.preprocessing import StandardScaler

from sklearn.ensemble import RandomForestRegressor

from sklearn.metrics import (
    r2_score,
    mean_absolute_error,
    mean_squared_error
)

from xgboost import XGBRegressor

from tensorflow.keras.models import Sequential

from tensorflow.keras.layers import Dense

from tensorflow.keras.optimizers import Adam

from tensorflow.keras.callbacks import EarlyStopping

# ==========================================================
# RANDOM SEED
# ==========================================================

SEED = 42

np.random.seed(SEED)

random.seed(SEED)

tf.random.set_seed(SEED)

print("Libraries Loaded Successfully")

# ==========================================================
# CELL 2: LOAD SP_GR DATASET
# ==========================================================

df = pd.read_csv(
    "sp_gr_dataset_dispersion.csv"
)

TARGET = "Hydrate Temperature (K)"

# ==========================================================
# FEATURES
# ==========================================================

X = df[
    [
        "Specific Gravity",
        "Pressure (psia)"
    ]
]

y = df[TARGET]

# ==========================================================
# GROUPING
# SAME GAS AT DIFFERENT PRESSURES
# ==========================================================

df["group_id"] = (
    df["Specific Gravity"]
    .round(4)
    .astype(str)
)

groups = df["group_id"]

# ==========================================================
# CELL 3: TRAIN TEST SPLIT
# ==========================================================

from sklearn.model_selection import GroupShuffleSplit

gss = GroupShuffleSplit(
    n_splits=1,
    test_size=0.20,
    random_state=42
)

train_idx, test_idx = next(
    gss.split(
        X,
        y,
        groups
    )
)

X_train = X.iloc[train_idx]
X_test  = X.iloc[test_idx]

y_train = y.iloc[train_idx]
y_test  = y.iloc[test_idx]

groups_train = groups.iloc[train_idx]

print("Train Samples :", len(X_train))
print("Test Samples  :", len(X_test))

print(
    "Train Groups :",
    groups_train.nunique()
)

print(
    "Test Groups :",
    groups.iloc[test_idx].nunique()
)
```

```python
# ==========================================================
# CELL 4: DEFINE BASE LEARNERS
# ==========================================================
from sklearn.ensemble import RandomForestRegressor
from sklearn.neural_network import MLPRegressor
from xgboost import XGBRegressor
from sklearn.ensemble import RandomForestRegressor
from xgboost import XGBRegressor

rf = RandomForestRegressor(
    n_estimators=200,
    max_depth=10,
    min_samples_split=5,
    min_samples_leaf=1,
    random_state=42,
    n_jobs=-1
)

xgb = XGBRegressor(
    n_estimators=500,
    learning_rate=0.05,
    max_depth=3,
    subsample=0.8,
    colsample_bytree=0.8,
    objective="reg:squarederror",
    random_state=42
)

# ==========================================================
# MLP
# ==========================================================

mlp = MLPRegressor(
    hidden_layer_sizes=(128, 64, 32),
    activation="relu",
    solver="adam",
    alpha=0.001,
    learning_rate_init=0.001,
    max_iter=2000,
    random_state=42
)

print("RF, XGB and MLP Initialized")

print("RF and XGB Initialized")
```

```python
# ==========================================================
# CELL 6: INITIALIZE GROUPKFOLD AND OOF ARRAYS
# ==========================================================
# Out-of-Fold predictions will be generated using
# GroupKFold to avoid data leakage.
# ==========================================================

from sklearn.model_selection import GroupKFold

gkf = GroupKFold(
    n_splits=5
)

# ==========================================================
# OOF ARRAYS
# ==========================================================

oof_rf = np.zeros(
    len(X_train)
)

oof_xgb = np.zeros(
    len(X_train)
)

oof_ann = np.zeros(
    len(X_train)
)

print(
    "Number of Training Samples:",
    len(X_train)
)

print(
    "Number of Groups:",
    groups_train.nunique()
)

print(
    "Number of Folds:",
    gkf.get_n_splits()
)

print(
    "OOF Arrays Created Successfully"
)
```

```python
# ==========================================================
# CELL 6: INITIALIZE OOF ARRAYS
# ==========================================================

oof_rf = np.zeros(len(X_train))

oof_xgb = np.zeros(len(X_train))

oof_mlp = np.zeros(len(X_train))

print("OOF Arrays Created")
print("Length =", len(oof_rf))
```

```python
# ==========================================================
# CELL 7: GENERATE OOF PREDICTIONS
# ==========================================================

for fold, (train_idx, valid_idx) in enumerate(

    gkf.split(
        X_train,
        y_train,
        groups_train
    )

):

    print("\n" + "="*60)
    print(f"FOLD {fold+1}")
    print("="*60)

    # ======================================================
    # SPLIT
    # ======================================================

    X_tr = X_train.iloc[train_idx]
    X_va = X_train.iloc[valid_idx]

    y_tr = y_train.iloc[train_idx]
    y_va = y_train.iloc[valid_idx]

    # ======================================================
    # RF
    # ======================================================

    rf.fit(
        X_tr,
        y_tr
    )

    rf_pred = rf.predict(
        X_va
    )

    oof_rf[valid_idx] = rf_pred

    # ======================================================
    # XGB
    # ======================================================

    xgb.fit(
        X_tr,
        y_tr
    )

    xgb_pred = xgb.predict(
        X_va
    )

    oof_xgb[valid_idx] = xgb_pred

    # ======================================================
    # MLP SCALING
    # ======================================================

    X_scaler = StandardScaler()

    X_tr_mlp = X_scaler.fit_transform(
        X_tr
    )

    X_va_mlp = X_scaler.transform(
        X_va
    )

    # ======================================================
    # MLP
    # ======================================================

    mlp.fit(
        X_tr_mlp,
        y_tr
    )

    mlp_pred = mlp.predict(
        X_va_mlp
    )

    oof_mlp[valid_idx] = mlp_pred

    # ======================================================
    # METRICS
    # ======================================================

    print(
        "RF R2 =",
        round(
            r2_score(
                y_va,
                rf_pred
            ),
            6
        )
    )

    print(
        "XGB R2 =",
        round(
            r2_score(
                y_va,
                xgb_pred
            ),
            6
        )
    )

    print(
        "MLP R2 =",
        round(
            r2_score(
                y_va,
                mlp_pred
            ),
            6
        )
    )
```

```python
# ==========================================================
# CELL 8: OOF RESULTS
# ==========================================================

results = pd.DataFrame({

    "Model": [
        "RF_OOF",
        "XGB_OOF",
        "MLP_OOF"
    ],

    "R2": [
        r2_score(y_train, oof_rf),
        r2_score(y_train, oof_xgb),
        r2_score(y_train, oof_mlp)
    ],

    "MAE": [
        mean_absolute_error(y_train, oof_rf),
        mean_absolute_error(y_train, oof_xgb),
        mean_absolute_error(y_train, oof_mlp)
    ],

    "RMSE": [
        np.sqrt(mean_squared_error(y_train, oof_rf)),
        np.sqrt(mean_squared_error(y_train, oof_xgb)),
        np.sqrt(mean_squared_error(y_train, oof_mlp))
    ]
})

print("\nOOF RESULTS\n")
print(results.sort_values(by="R2", ascending=False))
```

```python
# ==========================================================
# TRAIN & SAVE BASE MODELS
# RF + XGB + MLP
# ==========================================================

from sklearn.preprocessing import StandardScaler
from sklearn.neural_network import MLPRegressor
import joblib

# ==========================================================
# RETRAIN RF ON FULL DATA
# ==========================================================

rf.fit(
    X,
    y
)

# ==========================================================
# RETRAIN XGB ON FULL DATA
# ==========================================================

xgb.fit(
    X,
    y
)

# ==========================================================
# RETRAIN MLP ON FULL DATA
# ==========================================================

X_scaler_final = StandardScaler()

X_full_scaled = X_scaler_final.fit_transform(
    X
)

final_mlp = MLPRegressor(
    hidden_layer_sizes=(128, 64, 32),
    activation="relu",
    solver="adam",
    alpha=0.001,
    learning_rate_init=0.001,
    max_iter=2000,
    random_state=42
)

final_mlp.fit(
    X_full_scaled,
    y
)

# ==========================================================
# SAVE RF
# ==========================================================

joblib.dump(
    rf,
    "SPGR10_RF_base.pkl"
)

# ==========================================================
# SAVE XGB
# ==========================================================

joblib.dump(
    xgb,
    "SPGR10_XGB_base.pkl"
)

# ==========================================================
# SAVE MLP
# ==========================================================

joblib.dump(
    final_mlp,
    "SPGR10_MLP_base.pkl"
)

# ==========================================================
# SAVE MLP SCALER
# ==========================================================

joblib.dump(
    X_scaler_final,
    "SPGR10_MLP_X_scaler.pkl"
)

print("\nBase Models Saved Successfully")
```
